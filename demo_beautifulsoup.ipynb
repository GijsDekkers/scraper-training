### 1. Laad webpagina en parse HTML

```python
# URL van de demo-pagina
url = "https://www.example.com"
# url = "https://www.creates.nl"

# HTML ophalen
response = requests.get(url)
soup = BeautifulSoup(response.text, "html.parser")

# Alle zichtbare tekst ophalen
text = soup.get_text(separator="\n", strip=True)
print(text)
```

### 2. Haal alle paragrafen (P) op en print pagina titel

```python
# Alle paragrafen ophalen
for p in soup.find_all("p"):
    print(p.get_text())

# De titel van de pagina ophalen
title = soup.title.string
print("Pagina titel:", title)
```

### 3. Elementen met specifieke attributen

```python
# URL van de demo-pagina
url = "https://books.toscrape.com/"

# HTML ophalen
response = requests.get(url)
soup = BeautifulSoup(response.text, "html.parser")

# Alle producten ophalen
for article in soup.find_all("article", class_="product_pod"):
    # Titel
    title = article.h3.a["title"]

    # Prijs
    price = article.find("p", class_="price_color").get_text(strip=True)

    print(f"Titel: {title}, Prijs: {price}")
```

### 4. Geavanceerde selectie met CSS-selectors

```python
# Alle artikelen selecteren
for article in soup.select("article.product_pod"):
    # Titel en link
    title_elem = article.select_one("h3 a")
    title = title_elem["title"]
    link = title_elem["href"]

    # Prijs
    price = article.select_one(".price_color").get_text(strip=True)

    # Beschikbaarheid
    availability = article.select_one(".instock.availability").get_text(strip=True)

    # Rating (uit class)
    rating_elem = article.select_one(".star-rating")
    rating = rating_elem["class"][1] if rating_elem else "Onbekend"

    print(f"Titel: {title}")
    print(f"Prijs: {price}")
    print(f"Beschikbaarheid: {availability}")
    print(f"Beoordeling: {rating}")
    print(f"Link: {link}")
    print("-" * 40)
```
